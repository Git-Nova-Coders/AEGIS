# AEGIS Phase 1 — Model Evaluation & Explainability
### Objective: Compare model benchmarks (ROC-AUC, PR-AUC, Brier score), display reliability calibration diagrams, and review SHAP feature contributions.


```python
import json
import pandas as pd
from IPython.display import Image, display
from pathlib import Path
```

## 1. Model Performance Benchmark Comparison Table


```python
with open('../../ml/reports/model_comparison.json', 'r') as f:
    metrics_data = json.load(f)

test_metrics = pd.DataFrame(metrics_data['test_metrics'])
cols_order = ['model_name', 'roc_auc', 'pr_auc', 'brier_score', 'f1_score', 'recall', 'accuracy']
display(test_metrics[cols_order].sort_values(by='brier_score'))
```
